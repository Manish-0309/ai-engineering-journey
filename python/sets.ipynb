```python
# Declaration of set, uses {}
s = {"Manish", 61, 85.6, 'a'}
s
```

```python
# add(): add element, unordered way, no indexing
s.add("Kumar")

s
```

```python
s[2]
# no support to index based access
```

```python
# using loop to add element in the set
for i in range(25):
    s.add(i)

s
```

```python
# from list to tuple
lst = ["Manish", "Kumar", 1, 2, 3, 4, 5, 5, 4, 3, 2]
unique = set(lst)

unique
```

```python
# Set supports operations like:
# Union, intersection
s | unique
```

```python
# Intersection
s & unique
```

```python
# Difference
s - unique
```

```python
# symmetric difference
s ^ unique
```
