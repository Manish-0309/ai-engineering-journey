```python
# map() : applies on every items in an iterable and returns a map object(lazy iterator)
# syntax
    # map(function, iterable)
```

```python
# PS: Print square of each item in a list and return the list of squares
# general function

l = [1,2,3,4,5]

def square_list(l):
    result = []
    for i in l:
        result.append(i**2)
    return result

print(square_list(l))
```

```python

print(list(map(lambda x: x**2, l)))
```

```python
# user defined functions, lambda functions, and built in functions can be passed as an argument to map() function. The function is applied to every item of the iterable and a list of results is returned.
```

```python
# map() functions with multiple iterables

num1 = [1,2,3,4,5]
num2 = [6,7,8,9,10]

print(list(map(lambda x,y : x + y, num1, num2)))
```

```python
# map() function can also be used to retrieve data from list of dictionaries.
characters = [
    {'name' : 'Luffy', 'age' : 19, 'status': 'Yonko'},
    {'name' : 'Zoro', 'age' : 21, 'status': 'Swordsman'},
    {'name' : 'Nami', 'age' : 20, 'status': 'Navigator'},
    {'name' : 'Usopp', 'age' : 19, 'status': 'Sniper'},
    {'name' : 'Sanji', 'age' : 21, 'status': 'Cook'},
]

print(list(map(lambda x: x['status'], characters)))
```
