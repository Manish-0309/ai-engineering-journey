```python
# Dictionary Declaration
student = {
    "name" : "Luffy",
    "age" :19,
    "grade": "God"
}

student
```

```python
# accessing using key name
student["name"]
```

```python
# adding new key
student['status'] = "emperor"
```

```python
student
```

```python
# updating existing key
student['grade'] = 'Nika'
```

```python
student
```

```python
# deleting key-value from dictionary
del student['grade']
```

```python
student
```

```python
# Different methods of Dictionary
# keys(): return all the keys
student.keys()
```

```python
# values(): return all the values
student.values()
```

```python
# items(): return key-value together in tuple format
student.items()
```

```python
# get(): return value of which key is passed
student.get("name")
```

```python
# pop(key): remove the key and return value of that key
student.pop('status')
```

```python
# clear(): empty the dictionary
student.clear()
```

```python
# Shallow copy creation of dictionary
# Way to create a copy of original dictionary in a way that chnages made in original dictionary does not be reflected in copy dictionary. Method is copy and called as Shallow copy

orig_dict = {
    'a':11,
    'b':18,
    'c':15,
    'd':12,
}

orig_dict
```

```python
# creating shallow copy
copy_dict = orig_dict.copy()
```

```python
copy_dict
```

```python
orig_dict['a'] = 45
```

```python
orig_dict
```

```python
copy_dict
```

```python
# nested dictionary: dictionary inside dictionary
# can be used to store student details Eg.
students = {
    "student1" : {"Name": "Zoro", "age" : 23},
    "student2" : {"Name" : "Robin", "age" : 33}
}

students
```

```python
for i in students:
    print(students[i])
```

```python
for key, value in students.items():
    print(key, value)
```

```python
# Dictionary comprehension: It is a method to create dictionary in one line of code. It is similar to list comprehension. It is used to create dictionary from iterable objects like list, tuple, set etc.
# syntax: {key:value for item in iterable}

squares = {x: x*x for x in range(6)}

squares
```

```python
# Invert a dictionary: It is a method to create a new dictionary by swapping keys and values of existing dictionary. It is used to create a new dictionary from existing dictionary.
# syntax: {value:key for key, value in dictionary.items()}

inverted_dict = {v: k for k, v in squares.items()}
inverted_dict
```

```python
# Merge two dictionaries: It is a method to create a new dictionary by merging two existing dictionaries. It is used to create a new dictionary from existing dictionaries.
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}
merged_dict = {**dict1, **dict2}
merged_dict
```
