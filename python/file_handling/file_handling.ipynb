```python
# Modes in Python to perform above mentioned tasks:
# r - read file
# w - write into file (overwrite the entire comtent)
# a- append into the file, from the end of the file
# r+ - first read then write into the file
# w+ - first write and then read
```

```python
# Opening and reading the content of file
file = open('about_files.txt', mode='r')
content = file.read()

print(content)
file.close()
```

```python
# writing into the file
fw = open('about_files.txt', 'w')
fw.write('write() will overwrite everything in the existing file.')
fw.close()
```

```python
fw_read = open('about_files.txt', 'r')
print(fw_read.read())
fw.close()
```

```python
# appending into the file
fa = open('about_files.txt', 'a')
fa.write('''Information can be stored in RAM(temporarily) and it can be save(permanently). To save a file we need a file system.

Type of file:
1. Text Files: Human readable ASCII or UTF-8 characters.
        eg. : .txt, .csv, .json

2. Binary Files: Store data in raw binary format(0s, 1s). Not human readable.
        eg. : .png, .jpeg, .mp3, .mp4, .mov, etc

Commaon tasks we can perform with files:
    open, read, write, close

Modes in Python to perform above mentioned tasks:
r - read file
w - write into file (overwrite the entire comtent)
a- append into the file, from the end of the file
r+ - first read then write into the file
w+ - first write and then read''')

fa.close()
```

```python
fa_read = open('about_files.txt', 'r')
print(fa_read.read())
fa.close()
```

```python
# with: can be used to open file as a object and there is no need to close the file explicitly using close() function.
with open('about_files.txt', 'r') as wf:
    print(wf.read())

# same can be done with other modes
```

```python
# writeline() : help us write in our file line by line
lines = ['First Line\n', 'Second Line\n', 'Third Line\n']

with open('writelines.txt', 'w') as wt:
    wt.writelines(lines)
```

```python
with open('writelines.txt', 'r') as wr:
    print(wr.read())
```

```python
# While working with Binary files we can make use of 'rb' to read binary file and 'wb' to write into the file.
```

```python
# PS: read from the one file and write content into another file
with open('about_files.txt', 'r') as cp:
    content = cp.read()

with open('writelines.txt', 'w') as cw:
    cw.write(content)

```
