```python
# A function is a named, reusable block of a code that performs a specific task when called.
# syntax:
# def function_name(parameters):
#     # function body
#     return value  # optional
```

```python
# In Python, function supports:
'''
    default parameters, single parameter, multiple parameters, variable-length parameters(*args, **kwargs)
    -----> positional arguments - *args, keyword arguments - **kwargs
'''
```

```python
# function example
def greet(name):
    return f"Hello, {name}!"

greeting = greet("Alice")
print(greeting)  # Output: Hello, Alice!

# if the argument will not be passed, then we can set a default value for the parameter. But if the argument is passed, then the default value will be overridden.

def greet_with_default(name="World"):
    return f"Hello, {name}!"

greeting_default = greet_with_default()
print(greeting_default)  # Output: Hello, World!

greeting_custom = greet_with_default("Bob")
print(greeting_custom)  # Output: Hello, Bob!
```

```python
# Function with multiple parameters
def add_numbers(a, b):
    return a +b

result = add_numbers(5,3)
print(result) # Output: 8
```

```python
# Function with variable-length parameters
# Positional arguments (*args)
def sum_numbers(*args):
    added_sum = 0
    for num in args:
        added_sum += num
    return added_sum

sum_result = sum_numbers(1, 2, 3, 4, 5)
print(sum_result)  # Output: 15
```

```python
# Function with keyword-arguments (**kwargs)
def info(**kwargs):
    for key,value in kwargs.items():
        print(f'{key} : {value}')

info(name='Luffy', age=25, status='Emperor')
```

```python
# positional and keyword variables together
def data(*args, **kwargs):
    for i in args:
        print(f'Positional Values:', i)
    for k, v in kwargs.items():
        print(f'{k} : {v}')

data(1,2,3,4,5, name='Shanks', status='Yonko')
```

```python
# Positional argument after keyword arguments will result in errors.
data(1,2,3,4,5, name='Shanks', status='Yonko', 56)
```

```python
# Recursive functions: A function when it calls itself called as recursive function.
# Must have its base case: A condition, when met function stops its execution.
```

```python
# Recursive function example
def factorial(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial(num -1)

print(factorial(5))
```
