```python
# working with ndarray in numpy
import numpy as np

np.__version__
```

```python
# 0D - Scalar 
print(np.array(10))
```

```python
# 1D - Single axis 
print(np.array([10, 20, 30, 40]))
```

```python
# 2D Two Dimensional- Rows and Columns
print(np.array([[5, 6],[7, 8]]))        
```

```python
# 3D Array
print(np.array([
    [
        [5, 6],[7, 8]
    ],
    [
        [5,8],
        [3,4]
    ]
    ]))
```

```python
# Array attributes : ndim, shape, size, dtype, itemsize
ar = np.array([
    [
        [5, 6],[7, 8]
    ],
    [
        [5,8],
        [3,4]
    ]
    ])

print(ar)
print(ar.ndim)
```

```python
ar.shape
```

```python
ar.size
```

```python
ar.dtype
```

```python
ar.itemsize
```

```python
arr = np.array([45,4.5,68,89,46,12,2.0])

print(arr)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
# built-in creators in python
zeros = np.zeros(10)  #1D
zeros_2D = np.zeros((2,3))
```

```python
print(zeros)
print(zeros_2D)
```

```python
# ones 
ones = np.ones(5)
ones_2D = np.ones((4,3))

print(ones)
print(ones_2D)
```

```python
# identity matrix
iden = np.eye(5)
print(iden)
```

```python
# fill value with
fill_value = np.full((2,3), 4)
fill_value
```

```python
# sequence generation
seq = np.arange(0, 12, 1, dtype=None)
print(seq)
```

```python
# evenly spaced sequence 
ev_seq = np.linspace(0,2, 8)
```

```python
ev_seq
```

```python
# zero_division error
zero_even_seq = np.linspace(0,2, 0)
```

```python
zero_even_seq
```
