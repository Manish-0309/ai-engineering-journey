### Task 1 - Discount Rules(if/elif/else)

```python
order_amount = int(input("Enter the order amount:"))

if not isinstance(order_amount, int):
    print("Please enter a valid integer for the order amount.")
    exit()

if  1000 <= order_amount <1500:
    discount = 0.07
elif 1500 <= order_amount < 2000:
    discount = 0.1
elif order_amount >= 2000:
    discount = 0.15
else:
    discount = 0

tax_rate = 0.05

subtotal = order_amount - (order_amount * discount)
tax = subtotal * tax_rate
total_amount = subtotal + tax

print(f"Order Amount: {order_amount:.2f}")
print(f"Discount: {order_amount * discount:.2f}")
print(f"Subtotal after discount: {subtotal:.2f}")
print(f"Tax: {tax:.2f}")
print(f"Total Amount: {total_amount:.2f}")
```

### Task 2: Process multiple Orders (for loop)

```python
orders = [1200, 2500, 800, 1750, 3000]

for order in orders:
    if not isinstance(order, int):
        print("Please enter a valid integer for the order amount.")
        exit()

    if  1000 <= order <1500:
        discount = 0.07
    elif 1500 <= order < 2000:
        discount = 0.1
    elif order >= 2000:
        discount = 0.15
    else:
        discount = 0

    tax_rate = 0.05

    subtotal = order - (order * discount)
    tax = subtotal * tax_rate
    total_amount = subtotal + tax

    print(f"Order Amount: {order:.2f}")
    print(f"Discount: {order * discount:.2f}")
    print(f"Subtotal after discount: {subtotal:.2f}")
    print(f"Tax: {tax:.2f}")
    print(f"Total Amount: {total_amount:.2f}")

    print("----------------------------------")

total_revenue = sum(total_amount for order in orders)
print(f"Total_revenue: {total_revenue:.2f}")
```
